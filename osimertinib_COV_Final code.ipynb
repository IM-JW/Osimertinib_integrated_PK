```python
########2024-04-02 REF3 BMI to WT __ FINAL  ##############

import pandas as pd
import numpy as np

# 전체 ID 수 설정
total_ids = 500

# 라인별 비율 설정 및 라인 목록 생성
line_ratios = {'1st line': 32.2, '2nd line': 54, '3rd line': 13.8}
line_options = np.random.choice(['1st line', '2nd line', '3rd line'], size=total_ids, p=[line_ratios['1st line'] / 100, line_ratios['2nd line'] / 100, line_ratios['3rd line'] / 100])

# 데이터 프레임 생성 및 라인 할당
df = pd.DataFrame({'ID': range(1, total_ids + 1), 'Line': line_options})

# 성별 할당 함수
def assign_sex(row):
    sex_ratios = {'1st line': {'male': 42.9, 'female': 57.1}, '2nd line': {'male': 27.7, 'female': 72.3}, '3rd line': {'male': 16.7, 'female': 83.3}}
    ratios = sex_ratios[row['Line']]
    return np.random.choice(['male', 'female'], p=[ratios['male']/100, ratios['female']/100])

df['Sex'] = df.apply(assign_sex, axis=1)

# 나이 할당 함수
def assign_age_correctly_v2(line):
    age_distributions = {'1st line': (63.0, 55.8, 72.2), '2nd line': (68.0, 56.5, 78.5), '3rd line': (69.5, 59.2, 73.0)}
    median_age, p25, p75 = age_distributions[line]
    age = np.random.uniform(p25, p75)
    return round(age, 1)

df['Age'] = df['Line'].apply(assign_age_correctly_v2)

# SEX_N 할당
df['SEX_N'] = df['Sex'].apply(lambda x: 1 if x == 'male' else 0)

# HT 값 할당 함수
def assign_ht(row):
    if row['Sex'] == 'male':
        if 50 <= row['Age'] < 60:
            return 175.5
        elif 60 <= row['Age'] < 70:
            return 174.4
        elif 70 <= row['Age'] < 80:
            return 173.0
    else:
        if 50 <= row['Age'] < 60:
            return 161.3
        elif 60 <= row['Age'] < 70:
            return 160.7
        elif 70 <= row['Age'] < 80:
            return 158.2
    return np.nan

df['HT'] = df.apply(assign_ht, axis=1)

# BMI 할당 함수
def assign_bmi(line):
    bmi_distributions = {'1st line': (22.0, 19.6, 24.4), '2nd line': (23.2, 21.0, 26.0), '3rd line': (21.1, 17.9, 23.9)}
    median_bmi, p25, p75 = bmi_distributions[line]
    bmi = np.random.uniform(p25, p75)
    return round(bmi, 1)

df['BMI'] = df['Line'].apply(assign_bmi)

# WT 값 계산 및 할당 함수
def calculate_wt(row):
    ht_in_meters = row['HT'] / 100  # 'HT' 칼럼의 값을 미터 단위로 변환
    wt = row['BMI'] * (ht_in_meters ** 2)  # BMI 공식에 따라 체중 계산
    return round(wt, 1)

df['WT'] = df.apply(calculate_wt, axis=1)

# CSV 파일로 저장
df.to_csv('/mnt/data/final_dataset_with_wt.csv', index=False)
```

```python
### REF3 Lab 수치 관련

import pandas as pd
import numpy as np

# 전체 ID 수 설정
total_ids = 87

# 라인별 비율 설정
line_ratios = {'1st line': 32.2, '2nd line': 54, '3rd line': 13.8}

# 라인별 비율에 따라 라인 목록 생성
line_options = ['1st line'] * int(total_ids * (line_ratios['1st line'] / 100)) + \
               ['2nd line'] * int(total_ids * (line_ratios['2nd line'] / 100)) + \
               ['3rd line'] * int(total_ids * (line_ratios['3rd line'] / 100))

# 실제 total_ids와 생성된 라인 옵션의 길이 차이를 조정
if len(line_options) > total_ids:
    line_options = line_options[:total_ids]
elif len(line_options) < total_ids:
    line_options += np.random.choice(['1st line', '2nd line', '3rd line'], size=total_ids-len(line_options), p=[line_ratios['1st line'] / 100, line_ratios['2nd line'] / 100, line_ratios['3rd line'] / 100]).tolist()

# 라인 선택을 무작위로 섞음
np.random.shuffle(line_options)

# 데이터 프레임 생성 및 라인 할당
df = pd.DataFrame({'ID': range(1, total_ids + 1), 'Line': line_options})

# 라인별 ALT, AST, ALB, TBB 값을 설정
def assign_values_extended(row):
    if row['Line'] == '1st line':
        row['ALT'] = np.random.uniform(14.5, 43.0)
        row['AST'] = np.random.uniform(18.0, 30.5)
        row['ALB'] = np.random.uniform(35.5, 42.8)
        row['TBB'] = np.random.uniform(6.0, 8.1)
    elif row['Line'] == '2nd line':
        row['ALT'] = np.random.uniform(17.5, 33.0)
        row['AST'] = np.random.uniform(22.0, 31.5)
        row['ALB'] = np.random.uniform(36.0, 40.0)
        row['TBB'] = np.random.uniform(4.1, 7.4)
    elif row['Line'] == '3rd line':
        row['ALT'] = np.random.uniform(17.5, 33.5)
        row['AST'] = np.random.uniform(18.0, 27.5)
        row['ALB'] = np.random.uniform(39.5, 42.5)
        row['TBB'] = np.random.uniform(5.5, 11.5)
    return row

# 새로운 'TBB' 값을 포함하여 다시 적용
df = df.apply(assign_values_extended, axis=1)

# 데이터셋을 CSV 파일로 저장 (이번에는 'TBB' 값을 포함하여)
file_path_extended = "/mnt/data/line_based_dataset_with_tbb.csv"
df.to_csv(file_path_extended, index=False)
```
