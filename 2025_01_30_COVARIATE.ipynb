```python
## ISHIKAWA

import numpy as np
import pandas as pd

# 설정된 총 인원
n = 450

# 1. Sex (Male=1, Female=0) - 39.6% / 60.4% 비율
sex = np.random.choice([1, 0], size=n, p=[0.396, 0.604])

# 2. AGE (Median [IQR]) -> 69.0 [60-74]
age_median = 69.0
age_iqr = [60, 74]
age_std = (age_iqr[1] - age_iqr[0]) / 1.35
age = np.random.normal(loc=age_median, scale=age_std, size=n)


# 3. Serum Creatinine (Median [IQR]) -> 73.4 [61.9–93.7]
creatinine_median = 73.4
creatinine_iqr = [61.9, 93.7]
creatinine_std = (creatinine_iqr[1] - creatinine_iqr[0]) / 1.35
serum_creatinine = np.random.normal(loc=creatinine_median, scale=creatinine_std, size=n)


# 4. Albumin (Median [IQR]) -> 40 [38-43]
albumin_median = 40
albumin_iqr = [38, 43]
albumin_std = (albumin_iqr[1] - albumin_iqr[0]) / 1.35
albumin = np.random.normal(loc=albumin_median, scale=albumin_std, size=n)


# 5. Body Weight (Median [IQR]) -> 55 [49-64]
weight_median = 55
weight_iqr = [49, 64]
weight_std = (weight_iqr[1] - weight_iqr[0]) / 1.35
weight = np.random.normal(loc=weight_median, scale=weight_std, size=n)


# 6. AST (Median [IQR]) -> 23 [19-28]
ast_median = 23
ast_iqr = [19, 28]
ast_std = (ast_iqr[1] - ast_iqr[0]) / 1.35
ast = np.random.normal(loc=ast_median, scale=ast_std, size=n)


# 7. ALT (Median [IQR]) -> 17 [12-23]
alt_median = 17
alt_iqr = [12, 23]
alt_std = (alt_iqr[1] - alt_iqr[0]) / 1.35
alt = np.random.normal(loc=alt_median, scale=alt_std, size=n)


# 8. ID 생성 (1부터 450까지)
id_col = np.arange(1, n+1)

# 데이터프레임 생성
df = pd.DataFrame({'ID': id_col, 'Sex': sex, 'Age': age, 'Serum_Creatinine': serum_creatinine,
                   'Albumin': albumin, 'Body_Weight': weight, 'AST': ast, 'ALT': alt})

# 데이터 랜덤 섞기
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# CSV 파일 저장
df.to_csv('ISHI_COV.csv', index=False)

# 데이터 확인
df.head()
```

```python
## Rodier


import numpy as np
import pandas as pd

# 설정된 총 인원
n = 150

# 1. Sex (Male=1, Female=0) - 54.0% / 46.0% 비율
sex = np.random.choice([1, 0], size=n, p=[0.54, 0.46])

# 2. BMI (Median [IQR]) -> 23.2 [21.0–26.0]
bmi_mean = 23.2
bmi_iqr = [21.0, 26.0]
bmi_std = (bmi_iqr[1] - bmi_iqr[0]) / 1.35  # IQR을 표준편차로 변환
bmi = np.random.normal(loc=bmi_mean, scale=bmi_std, size=n)


# 3. Age (Median [IQR]) -> 68.0 [56.5-78.5]
age_median = 68.0
age_iqr = [56.5, 78.5]
age_std = (age_iqr[1] - age_iqr[0]) / 1.35
age = np.random.normal(loc=age_median, scale=age_std, size=n)


# 4. Serum Creatinine (Median [IQR]) -> 82.0 [70.0-94.0]
creatinine_median = 82.0
creatinine_iqr = [70.0, 94.0]
creatinine_std = (creatinine_iqr[1] - creatinine_iqr[0]) / 1.35
serum_creatinine = np.random.normal(loc=creatinine_median, scale=creatinine_std, size=n)


# 5. Albumin (Median [IQR]) -> 38.0 [36.0-40.0]
albumin_median = 38.0
albumin_iqr = [36.0, 40.0]
albumin_std = (albumin_iqr[1] - albumin_iqr[0]) / 1.35
albumin = np.random.normal(loc=albumin_median, scale=albumin_std, size=n)


# 6. AST (Median [IQR]) -> 26.0 [22.0-31.5]
ast_median = 26.0
ast_iqr = [22.0, 31.5]
ast_std = (ast_iqr[1] - ast_iqr[0]) / 1.35
ast = np.random.normal(loc=ast_median, scale=ast_std, size=n)


# 7. ALT (Median [IQR]) -> 26.0 [17.5-33.0]
alt_median = 26.0
alt_iqr = [17.5, 33.0]
alt_std = (alt_iqr[1] - alt_iqr[0]) / 1.35
alt = np.random.normal(loc=alt_median, scale=alt_std, size=n)

# 8. ID 생성 (1부터 150까지)
id_col = np.arange(1, n+1)

# 데이터프레임 생성
df = pd.DataFrame({'ID': id_col, 'Sex': sex, 'BMI': bmi, 'Age': age,
                   'Serum_Creatinine': serum_creatinine, 'Albumin': albumin,
                   'AST': ast, 'ALT': alt})

# 연령 그룹별 키 설정 함수
def get_height(row):
    age, sex = row['Age'], row['Sex']
    if 20 <= age < 30:
        return 1.758 if sex == 1 else 1.626
    elif 30 <= age < 40:
        return 1.763 if sex == 1 else 1.627
    elif 40 <= age < 50:
        return 1.763 if sex == 1 else 1.623
    elif 50 <= age < 60:
        return 1.754 if sex == 1 else 1.613
    elif 60 <= age < 70:
        return 1.744 if sex == 1 else 1.607
    elif 70 <= age < 80:
        return 1.730 if sex == 1 else 1.582
    elif 80 <= age <= 90:
        return 1.704 if sex == 1 else 1.556
    return 1.7 if sex == 1 else 1.6  # 기본값

# 각 연령 및 성별에 따라 키 설정
df['Height'] = df.apply(get_height, axis=1)

# 체중(Weight) 계산: Weight = BMI * Height^2
df['Weight'] = df['BMI'] * (df['Height'] ** 2)

# WT 계산: sqrt(Weight)
df['WT'] = np.sqrt(df['Weight'])

# 데이터 랜덤 섞기
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# CSV 파일 저장
df.to_csv('Rodier_COV.csv', index=False)

# 데이터 확인
df.head()
```
